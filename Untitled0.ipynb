```python
# ============================================================
# CODSOFT DATA ANALYTICS INTERNSHIP - TASK 1, 2, 3
# Dataset: Seaborn's built-in "tips" dataset (restaurant sales data)
# No download needed — loads directly from the internet via seaborn
# ============================================================

# ---- SETUP (run this first in Colab) ----
# pip install seaborn pandas matplotlib is usually already available in Colab
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# ============================================================
# TASK 1: DATA CLEANING & PREPROCESSING
# ============================================================

# Import the dataset
df = sns.load_dataset("tips")

# Inspect structure
print("Shape (rows, columns):", df.shape)
print("\nColumn info:")
print(df.info())
print("\nFirst 5 rows:")
print(df.head())

# Identify missing values
print("\nMissing values per column:")
print(df.isnull().sum())

# Identify duplicate records
print("\nNumber of duplicate rows:", df.duplicated().sum())

# Clean the dataset: drop duplicates (if any), handle nulls (if any)
df_clean = df.drop_duplicates()
df_clean = df_clean.dropna()

# Correct data types (example: making sure 'size' is integer)
df_clean["size"] = df_clean["size"].astype(int)

# Check final structure after cleaning
print("\nShape after cleaning:", df_clean.shape)

# BONUS: Save cleaned dataset as a new CSV file
df_clean.to_csv("cleaned_tips_data.csv", index=False)
print("\nCleaned dataset saved as cleaned_tips_data.csv")
```

```python
# ============================================================
# TASK 2: EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================

# Descriptive statistics
print("\n--- Descriptive Statistics ---")
print(df_clean.describe())

# Trends & relationships: correlation between numeric variables
print("\n--- Correlation Matrix ---")
print(df_clean.corr(numeric_only=True))

# Distribution check: average bill and tip by day
print("\n--- Average total_bill and tip by day ---")
print(df_clean.groupby("day")[["total_bill", "tip"]].mean())

# Outlier detection using IQR method on total_bill
Q1 = df_clean["total_bill"].quantile(0.25)
Q3 = df_clean["total_bill"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = df_clean[(df_clean["total_bill"] < lower_bound) | (df_clean["total_bill"] > upper_bound)]
print(f"\nNumber of outliers in total_bill: {len(outliers)}")
print(outliers)

# Answering business questions using summary stats
print("\n--- Business Questions ---")
print("Q1: Which day has the highest average bill?")
print(df_clean.groupby("day")["total_bill"].mean().idxmax())

print("\nQ2: Do smokers tip more than non-smokers on average?")
print(df_clean.groupby("smoker")["tip"].mean())

print("\nQ3: Does group size affect the bill amount?")
print(df_clean.groupby("size")["total_bill"].mean())

# BONUS: Short report (printed summary — copy this into a Word/text report)
report = f"""
DATA ANALYSIS REPORT - Restaurant Tips Dataset
------------------------------------------------
1. Dataset contains {df_clean.shape[0]} records after cleaning.
2. The busiest day by average bill is {df_clean.groupby("day")["total_bill"].mean().idxmax()}.
3. Average tip percentage overall is {(df_clean['tip'].sum() / df_clean['total_bill'].sum() * 100):.2f}%.
4. {len(outliers)} outlier records were found in total_bill using the IQR method.
5. Larger group sizes tend to have higher total bills, as expected.
"""
print(report)
```

```python
# ============================================================
# TASK 3: DATA VISUALIZATION DASHBOARD
# ============================================================

sns.set_style("whitegrid")

# 1. Bar chart: Average total_bill by day
plt.figure(figsize=(6,4))
sns.barplot(data=df_clean, x="day", y="total_bill", hue="day", palette="viridis", legend=False)
plt.title("Average Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Average Total Bill ($)")
plt.tight_layout()
plt.savefig("chart1_bar_bill_by_day.png")
plt.show()

# 2. Line chart: Average tip by group size
plt.figure(figsize=(6,4))
tip_by_size = df_clean.groupby("size")["tip"].mean().reset_index()
sns.lineplot(data=tip_by_size, x="size", y="tip", marker="o", color="darkorange")
plt.title("Average Tip by Group Size")
plt.xlabel("Group Size")
plt.ylabel("Average Tip ($)")
plt.tight_layout()
plt.savefig("chart2_line_tip_by_size.png")
plt.show()

# 3. Pie chart: Smoker vs Non-smoker proportion
plt.figure(figsize=(5,5))
df_clean["smoker"].value_counts().plot.pie(autopct="%1.1f%%", colors=["#66b3ff","#ff9999"], startangle=90)
plt.title("Smoker vs Non-Smoker Customers")
plt.ylabel("")
plt.tight_layout()
plt.savefig("chart3_pie_smoker.png")
plt.show()

# 4. Histogram: Distribution of total_bill
plt.figure(figsize=(6,4))
sns.histplot(df_clean["total_bill"], bins=20, kde=True, color="mediumseagreen")
plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill ($)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("chart4_hist_total_bill.png")
plt.show()

# 5. Scatter plot: Total bill vs Tip
plt.figure(figsize=(6,4))
sns.scatterplot(data=df_clean, x="total_bill", y="tip", hue="time", palette="Set2")
plt.title("Total Bill vs Tip Amount")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.legend(title="Time")
plt.tight_layout()
plt.savefig("chart5_scatter_bill_vs_tip.png")
plt.show()

print("\nAll charts saved as PNG files. You can download them from the Colab file panel.")
```
